```python

import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kavin", "Divya", "Rahul"],
    "Age": [21, 22, 20, 23, 21],
    "Course": ["MCA", "BCA", "MCA", "B.Sc", "MCA"],
    "Marks": [85, 92, 78, 88, 95]
}

df = pd.DataFrame(data)

print("Complete Dataset")
print(df)

print("\nSelect rows using loc")
print(df.loc[0:2, ["Name", "Marks"]])

print("\nSelect rows using iloc")
print(df.iloc[0:3, 0:2])

print("\nSelect specific columns")
print(df.loc[:, ["Name", "Course"]])

print("\nStudents scoring 85 or above")
print(df.loc[df["Marks"] >= 85, ["Name", "Marks"]])
```
